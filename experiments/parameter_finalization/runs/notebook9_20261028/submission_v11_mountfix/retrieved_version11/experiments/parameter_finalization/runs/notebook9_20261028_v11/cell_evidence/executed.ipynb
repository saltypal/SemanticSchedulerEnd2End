```
print('START Qwen3.5 labeling of300 images',flush=True)
if RUN_REAL:
 from parameters.dataset_pipeline import label_dataset,export_spi_training,load_selected_teacher,materialize_current_300
 if CONFIG.get('input_root') and CONFIG.get('records_path'):raise ValueError('Choose input_root or records_path, not both')
 if CONFIG.get('input_root'):
  records,features=materialize_current_300(CONFIG['input_root'],OUTPUT/'input_images')
  CONFIG['features_path']=str(features)
 else:records=json.loads(Path(CONFIG['records_path']).read_text())
 if len(records)!=300:raise ValueError('This milestone requires exactly300 images')
 # Fail early on wrong feature space or broken detector before 300 VLM calls.
 from parameters.minimal_vision import MinimalVisionProvider
 from parameters.contracts import MissionSpec
 from parameters.extractor import embedding_signature
 from parameters.spatial_provider import make_spatial_provider
 from PIL import Image
 import numpy as np,gc,torch
 preflight_mission=MissionSpec('preflight','Describe visible evidence.',(('visible','a photograph'),))
 preflight_provider=MinimalVisionProvider(CONFIG['clip_checkpoint'],{'preflight':preflight_mission},device=CONFIG.get('device','cpu'))
 preflight_scene=make_spatial_provider(CONFIG['detector_checkpoint'],detector=CONFIG['detector'],device=CONFIG.get('device','cpu'))
 try:
  with Image.open(records[0]['image_path']) as im: preflight_image=im.copy()
  preflight_evidence=preflight_provider.extract_batch([preflight_image],[preflight_mission])[0]
  with np.load(CONFIG['features_path'],allow_pickle=False) as archive: expected_signature=str(archive['feature_signature'].item())
  if embedding_signature(preflight_evidence)!=expected_signature:raise RuntimeError('Fresh CLIP producer differs from cached training feature space')
  preflight_measurements=preflight_scene.measure(preflight_image)
  if preflight_measurements.status!='ok':raise RuntimeError('Detector preflight failed: '+str(preflight_measurements.reason))
  from parameters.artifacts import write_json
  write_json(OUTPUT/'producer_preflight.json',{'feature_signature':expected_signature,'scene':preflight_measurements.audit,'status':'passed'})
 finally:
  preflight_provider.close()
  del preflight_provider,preflight_scene
  gc.collect()
  if torch.cuda.is_available():torch.cuda.empty_cache()
 print('Producer preflight passed; starting accepted teacher',flush=True)
 teacher=load_selected_teacher(TEACHER_RECEIPT)
 try:label_dataset(records,teacher,TEACHER_RECEIPT,OUTPUT/'teacher_labels')
 finally:
  teacher.close()
  del teacher
  gc.collect()
  if torch.cuda.is_available():torch.cuda.empty_cache()
 export_spi_training(CONFIG['features_path'],OUTPUT/'teacher_labels/labels.json',OUTPUT/'spi_training')
 print('Teacher labels and training bundle saved',flush=True)
else:print('Prepared only: no real labeling launched',flush=True)
print('END Qwen3.5 labeling stage',flush=True)
```

```
print('START four-class SPI classifier training on300-image cohort',flush=True)
if RUN_REAL:
 from parameters.dataset_pipeline import train_spi_for_cache
 from parameters.artifacts import write_json
 TRAINING=train_spi_for_cache(OUTPUT/'spi_training/features.npz',OUTPUT/'spi_training/labels.json',
   TEACHER_RECEIPT,OUTPUT/'spi_classifier',ROOT/'parameters/config/parameter_release_v2.json',root=ROOT,seconds=600)
 write_json(OUTPUT/'spi_training_result.json',TRAINING)
 print('Student validation',TRAINING['validation'],flush=True)
 print('Parameter cache permitted',TRAINING['cache_allowed'],flush=True)
 if not TRAINING['cache_allowed']:raise RuntimeError('Student support/recall/export checks failed; preserve model, stop before cache')
 CONFIG['model_path']=TRAINING['model_path'];CONFIG['readiness_path']=TRAINING['readiness_path']
else:print('Prepared only: no real SPI classifier trained',flush=True)
print('END four-class SPI training',flush=True)
```

```
print('START readiness-gated parameter cache',flush=True)
if RUN_REAL:
 from parameters.readiness import require_spi_ready
 from parameters.dataset_pipeline import cache_parameters
 from parameters.minimal_vision import MinimalVisionProvider
 from parameters.contracts import MissionSpec
 from parameters.spatial_provider import make_spatial_provider
 require_spi_ready(CONFIG['model_path'],CONFIG['readiness_path'])
 p=json.loads((ROOT/'parameters/config/ue_parameter_profiles_v2.json').read_text())
 profiles=[(r['ue_id'],MissionSpec(r['mission_id'],r['paragraph'],((r['mission_id'],r['paragraph']),),allowed_delay_ms=r['allowed_delay_ms'])) for r in p['profiles']]
 provider=MinimalVisionProvider(CONFIG['clip_checkpoint'],{m.mission_id:m for _,m in profiles},device=CONFIG.get('device','cpu'))
 scene=make_spatial_provider(CONFIG['detector_checkpoint'],detector=CONFIG.get('detector','yolo_world'),device=CONFIG.get('device','cpu'))
 try:cache_parameters(records,profiles,provider,scene,CONFIG['model_path'],CONFIG['readiness_path'],OUTPUT/'parameter_cache')
 finally:
  provider.close()
  if hasattr(scene,'close'):scene.close()
  del scene,provider
  gc.collect()
  if torch.cuda.is_available():torch.cuda.empty_cache()
else:print('BLOCKED: real four-class SPI readiness is pending; dummy cache is separate.',flush=True)
print('END parameter-cache stage',flush=True)
```
